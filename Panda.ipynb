# Series data Structure

The series structure is one of the core data structures in pandas. The items are all stored in an order
and there's labels with which you can retrieve them. 
Two columns of data: the first is the speical index, like key in dictionary. the second is the actual data.
The data column has a label of its own and can be retrieved using the .name attribute.

```python
import pandas as pd

#create a series by passing in a list of values.
#when you do this Pandas automatically assigns an index starting with zero and sets the name of the series to None.

#create an arrar-like object, like a list
students = ['Alice', 'Jack', 'Molly'] #make a list first
#turn that into a series
pd.Series(students)
```

```python
#the result is a Series object which is nicely rendered to the screen. 
#pandas automatically identified the type of data in this series as 'Object'

#if we passed in a list of whole numbers
numbers = [1, 2, 3]
#and turn that into a series
pd.Series(numbers)
```

```python
#how numpy and pandas handle missing data
#we have the none type to indicate a lack of data

students = ['Alice', 'Jack', None]
#And lets convert this to a series
pd.Series(students)
```

```python
#if we create a list of numbers, integers or floats, and put in the None type
#pandas automatically converts this to a special floating point value designed as NaN
numbers = [1, 2, None]
#And lets convert this to a series
pd.Series(numbers)
```

```python
#'NaN' which stands for 'not a number', it is a special floating point value
#pandas went on and converted the integers to floats.
#when the list of integers you put into a series is not floats, it's probabily because there is some missing data.
```

```python
#NaN is not equivalent to None
import numpy as np
np.nan == None
```

```python
#you cannot do an equality test of NAN to itself, if you do, the answer is always false
np.nan == np.nan
```

```python
#you need to use some special functions to test for the presence of not a number
#such as the Numpy library isnan()

np.isnan(np.nan)
```

```python
#when you see NaN, it's meaning is similar to None, but it's a numeric value and treated differently for
#efficiency reasons
```

```python
#a series can be created directly from dictionary data
#the index is automatically assigned to the keys of the dictionary you provided

students_score = {'Alice': 'Physics', 
                 'Jack': 'Chemistry', 
                 'Molly': 'English'}
s = pd.Series(students_score)
s
```

# Querying series

The primary data types of the panda library, the Series.

A pandas Series can be queried either by the index position or the index label.
To query by numeric location, starting at zero, use the iloc attribute. To query by the index label, you can use the loc attribute.

```python
import pandas as pd
#use students enrolled in classes coming from a dictionary
students_classes = {'Alice': 'Physics', 
                   'Jack': 'Chemistry', 
                   'Molly': 'English', 
                   'Sam': 'History'}
s = pd.Series(students_classes)
s
```

```python
#to see the fourth entry, we would use the iloc attribute with the parameter 3

s.iloc[3]

#remember we start from zero
```

```python
#to see what class Molly has, we can use the loc attribute with a parameter of Molly

s.loc['Molly']
```

```python
#iloc and loc are not methods, they are attributes. so use [] to query them, which is called the index operator
```

```python
#Pandas provides a sort of smart syntax using the indexing operator directly on the series itself.
#for example, if you pass in an integer parameter, the operator will behave as if you want it to query via the iloc
#attribute
```

```python
s[3] #s.iloc[3]
```

```python
#input an object, it will query as if you want to use the label based loc attribute
s['Molly'] #s.loc['Mooly']
```

```python
#index is a list of integers
#Pandas can't determine automaticallly whether you're intending to query by index position or index label.
#safer option would be use the iloc or loc attributes directly

class_code = {99: 'Physics', 
             100: 'Chemistry', 
             101: 'English', 
             102: 'History'}
s = pd.Series(class_code)

#s[0] will raise a key error because there's no item in the classes list with an index of zero

s.iloc[0]
```

```python
#iterate all the items in the series

grades = pd.Series([90, 80, 70, 60]) #pass a list into series

total = 0
for grade in grades:
    total = total + grade
print(total/len(grades))
```

```python
#This works but it's slow.
#Pandas and numpy support a method of computation called vectorization.
#Vectorization works with most of the functions in the numpy library, including the sum function.
```

```python
#use the numpy sum method
import numpy as np

#Then we just call np.sum and pass in an iterable item. In this case, our panda series.

total = np.sum(grades)
print(total/len(grades))
```

```python
#Which one is faster?
#First create a big series of random numbers
```

```python
numbers = pd.Series(np.random.randint(0, 1000, 10000))
numbers.head() 
#look at top five items in the series to make sure they actually seem random
#we do this with the head() function
```

```python
#verify the length of the series is correct using len function
len(numbers)
```

```python
#celluar magic function timeit
#run the code a few times to determine on average how long it takes
```

```python
#Now let's try it with vectorization
```

```python
%%timeit -n 100
total = np.sum(numbers)
total/len(numbers)
```

```python
#speed up in vectorization
```

```python
#broadcasting
#apply an operation to every value in the series, changing the series.

#look at the head of the series
numbers.head()
```

```python
# increase everything in the series by 2
numbers = numbers + 2
numbers.head()
```

```python
#The procedural way of doing this would be to iterate through all of the items in the series and increase the 
#values directly.

#We can use the iteritems() function which returns a label and value

for label, value in numbers.iteritems():
    #now for the item which is returned, lets call set_value
    numbers.at(label, value + 2)
#check the result of this computation
numbers.head()
```

```python
#speed comparisons
```

```python
%%timeit -n 10
s = pd.Series(np.random.randint(0, 1000, 1000))
for label, value in s.iteritems():
    s.loc[label] = value + 2
```

```python
#try that using the broadcasting methods
```

```python
%%timeit -n 10
s = pd.Series(np.random.randint(0, 1000, 1000))
s+=2 #using broadcasting
```

```python
#significantly faster, more concise and easier to read too
```

# Data frame data structure

The data frame is conceptually a two-dimensional series object, where there's an index and multiple columns of content, with each column having a label. The distinction between a column and a row is really only a conceptual distinction. And you can think of the DataFrame itself as simply a two-axes labelled array.

```python
import pandas as pd
record1 = pd.Series({'Name': 'Alice', 
                    'Class': 'Physics',
                    'Score': 85})
record2 = pd.Series({'Name': 'Jack', 
                    'Class': 'Chemistry',
                    'Score': 82})
record3 = pd.Series({'Name': 'Helen', 
                    'Class': 'Biology',
                    'Score': 90})
```

Like a series, the DataFrame object is index. Just like the series function, we can pass in our individual items in an array, and we can pass in our index values as a second arguments

```python
df = pd.DataFrame([record1, record2, record3],
                  index = ['school1', 'school2', 'school1'])
#use the head function to see the first several row of the dataframe, including indices from both axes

df.head()
```

```python
#index is the leftmost column and is the school name.
# we have rows of data, which each row has a column header which was given in our initial record dictionaries.
```

```python
#an alternative method is that you could use a list of dictionaries, where each dictionary represents a row of data
students = [{'Name': 'Alice', 
             'Class': 'Physics',
             'Score': 85}, 
            {'Name': 'Jack', 
             'Class': 'Chemistry',
             'Score': 82}, 
            {'Name': 'Helen', 
             'Class': 'Biology',
             'Score': 90} ]

#We pass the list of dictionaries into the DataFrame function

df = pd.DataFrame(students, index = ['school1', 'school2', 'school1'])
df.head()
```

```python
#We can extract data using the iloc and loc attributes
#Because it is two-dimensional, passing a single value to the loc indexing operator will return the series if there
#is only one row to return

#select data associated to school2, we just query the .loc attribute with one parameter

df.loc['school2']
```

```python
type(df.loc['school2'])
```

```python
#and we can see that the type is different too
type(df.loc['school1'])
```

```python
#dataframe - you can quickly select data based on multiple axes
#supply two parameters to .loc, one being the row index and the other being the column name

df.loc['school1', 'Name']
```

```python
#if we want to select a single column
#first, we could transpose the matrix, this pivots all of the rows into columns and all of the columns into rows
#it is done with the T attribute

df.T
```

```python
#Then we can call .loc on the transpose to get the students names only

df.T.loc['Name']
```

```python
#iloc and loc are used for row selection. Panda reserves the indexing operator directly on the DataFrame for column
#selection. So this selection is always label based, and is not as confusing as it was when using the square
#bracket operator on the series objects.
```

# Querying a DataFrame

The first step in the process is to understand Boolean masking.
Boolean masking is the heart of fast and efficient querying in numpy and pandas

```python
import pandas as pd 
df = pd.read_csv('E:\\Academics\\python\\applied data science in python\\introduction to data science to python\\week 2\\Admission_Predict.csv', index_col = 0)
df.columns = [x.lower().strip() for x in df.columns]
df.head()
```

```python
#Boolean masks are created by applying operators directly to the pandas series or dataframe objects

#To build a Boolean mask for this query
admit_mask = df['chance of admit'] > 0.7
admit_mask
```

```python
# So, what do you do with the boolean mask once you have formed it? Well, you can just lay it on top of the
# data to "hide" the data you don't want, which is represented by all of the False values. We do this by using
# the .where() function on the original DataFrame.
df.where(admit_mask).head()
```

```python
#Only data which met the condition was retained. All of the rows which did not meet the condition have NaN data instead
#If we don't want the NaN data, we use dropna() function

df.where(admit_mask).dropna().head()
```

```python
#a shorthead syntax which combines where() and dropna(), doing both at once
df[df['chance of admit'] >0.7].head()
```

```python
#called with a string parameter to project a single column

df['gre score'].head()
```

```python
#Or you can send it a list of columns as strings

df[['gre score', 'toefl score']].head()
```

```python
#or you can send it a boolean mask

df[df['gre score']>320].head()
```

```python
#combining multiple boolean masks, such as multiple criteria for including
(df['chance of admit']>0.7) and (df['chance of admit']<0.9)
#produce a traceback
```

```python
(df['chance of admit']>0.7) & (df['chance of admit']<0.9)
```

```python
#get rid of the comparison operator completely, and instead use the built in function

df['chance of admit'].gt(0.7) & df['chance of admit'].lt(0.9)
```

```python
df['chance of admit'].gt(0.7).lt(0.9)
```
