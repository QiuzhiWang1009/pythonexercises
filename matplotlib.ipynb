```python
import numpy as np
linear_data = np.array([1, 2, 3, 4, 5, 6, 7, 8])
quadratic_data = linear_data**2

plt.figure()
plt.plot(linear_data, '-o', quadratic_data, '-o')
plt.plot([22, 44, 55], '--r')

plt.xlabel('some data')
plt.ylabel('some other data')
plt.title('A title')
plt.legend(['Baseline', 'Competition', 'Us'])

plt.gca().fill_between(range(len(linear_data)), 
                    linear_data, quadratic_data, 
                    facecolor = 'blue', 
                    alpha = 0.5)
```

# Dejunkifying a plot

```python
import matplotlib.pyplot as plt
import numpy as np 

plt.figure()

languages = ['Python', 'SQL', 'Java', 'C++', 'JavaScript']
pos = np.arange(len(languages))
popularity = [56, 39, 34, 34, 29]

plt.bar(pos, popularity, align = 'center')
plt.xticks(pos, languages)
plt.ylabel('% Popularity')
plt.title('Top 5 languages for Math & data \n by % popularity on Stack Overflow', alpha = 0.8)

#to do: remove all the ticks (both axes) and tick labels on the Y axes

plt.tick_params(top = 'off', bottom= 'off', left = 'off', right = 'off', labelleft = 'off', labelbottom = 'on')

#to do: remove the frame of the chart

for spine in plt.gca().spines.values():
    spine.set_visible(False)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np 

plt.figure()

languages = ['Python', 'SQL', 'Java', 'C++', 'JavaScript']
pos = np.arange(len(languages))
popularity = [56, 39, 34, 34, 29]

#to do: change the bar color to be less bright blue
bars = plt.bar(pos, popularity, align = 'center', linewidth = 0, color = 'lightslategrey')
#to do: make one bar, the python bar, a contrasting color
bars[0].set_color('#1F77B4')

plt.bar(pos, popularity, align = 'center')

#to do: soften all labels by turning grey

plt.xticks(pos, languages, alpha = 0.8)

#to do: remove the Y label since bars are directly labeled
#plt.ylabel('% Popularity', alpha = 0.8)
plt.title('Top 5 languages for Math & data \n by % popularity on Stack Overflow', alpha = 0.8)

#to do: remove all the ticks (both axes) and tick labels on the Y axes

plt.tick_params(top = 'off', bottom= 'off', left = 'off', right = 'off', labelleft = 'off', labelbottom = 'on')

#to do: remove the frame of the chart

for spine in plt.gca().spines.values():
    spine.set_visible(False)
#to do: direct label each bar with Y axis values
for bar in bars: 
    plt.gca().text(bar.get_x() + bar.get_width()/2, bar.get_height() - 5, str(int(bar.get_height())) + '%', 
                   ha = 'center', color = 'w', fontsize = 11)
plt.show()
```

# Subplots

```python
import matplotlib.pyplot as plt
import numpy as np 
```

```python
plt.figure()
plt.subplot(1, 2, 1) #a subplot with one row and two columns
#first argument: number of rows, second argument: number of columns, third argument: plot number
linear_data = np.array([1, 2, 3, 4, 5, 6, 7, 8])

plt.plot(linear_data, '-o')
exponential_data = linear_data**2
plt.subplot(1, 2, 2) #plot on the second graph
plt.plot(exponential_data, '-o')
plt.subplot(1, 2, 1)
plt.plot(exponential_data, '-x')
```

```python
#lock one axis
plt.figure()
axl = plt.subplot(1, 2, 1)
plt.plot(linear_data, '-o')
ax2 = plt.subplot(1, 2, 2, sharey = axl) #share the y axis
plt.plot(exponential_data, '-x')
```

# Histograms

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, sharex = True)
axs = [ax1, ax2, ax3, ax4]

for n in range(0, len(axs)):
    sample_size = 10**(n+1)
    sample = np.random.normal(loc = 0.0, scale = 1.0, size = sample_size)
    axs[n].hist(sample)
    axs[n].set_title('n={}'.format(sample_size))
```

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, sharex = True)
axs = [ax1, ax2, ax3, ax4]
for n in range(0, len(axs)):
    sample_size = 10**(n+1)
    sample = np.random.normal(loc = 0.0, scale = 1.0, size = sample_size)
    axs[n].hist(sample, bins = 100) #change the bins to 100
    axs[n].set_title('n={}'.format(sample_size))
```

# Box and Whisker plots

```python
import pandas as pd
normal_sample = np.random.normal(loc = 0.0, scale = 1.0, size = 10000)
random_sample = np.random.random(size = 10000)
gamma_sample = np.random.gamma(2, size = 10000)

df = pd.DataFrame({'normal': normal_sample, 'random': random_sample, 'gamma': gamma_sample})
#use pandas described function to see some summary statistics about the data frame
df.describe()
plt.figure() 
_ = plt.boxplot(df['normal'], whis = 'range') # variable name is just the underscore
```

```python
plt.clf()
_ = plt.boxplot([df['normal'], df['random'], df['gamma']], whis = 'range')
```

```python
plt.figure()
_ = plt.hist(df['gamma'], bins = 100)
```

```python
import mpl_toolkits.axes_grid1.inset_locator as mpl_il 
plt.figure()
plt.boxplot([df['normal'], df['random'], df['gamma']], whis = 'range')
ax2 = mpl_il.inset_axes(plt.gca(), width = '60%', height= '40%', loc = 2)
ax2.hist(df['gamma'], bins = 100)
ax2.margins(x = 0.5)
ax2.yaxis.tick_right()
```

# Animations

```python
import matplotlib.animation as animation
import numpy as np
n = 100
x = np.random.randn(n)
```

```python
def update(curr):
    if curr == n:
        a.event_source.stop()
    plt.cla()
    bins = np.arange(-4, 4, 0.5)
    plt.hist(x[:curr], bins = bins)
    plt.axis([-4, 4, 0, 30])
    plt.gca().set_title('Sampling the Normal Distribution')
    plt.gca().set_ylabel('Frequency')
    plt.gca().set_xlabel('Value')
    plt.annotate('n = {}'.format(curr), [3, 27])
```

```python
fig = plt.figure()
a = animation.FuncAnimation(fig, update, interval = 100)
```
